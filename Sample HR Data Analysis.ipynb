```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
##pd.set_option('display.max_rows', None)
```

```python
data = pd.read_csv('Sample_HR.csv')
```

```python
data.head()
```

```python
data.shape
```

```python
data.count()
```

```python
## Find duplicate rows
data[data.duplicated()]
```

```python
## Find duplicate row counts
data[data.duplicated()].shape[0]
```

```python
## Remove duplicate rows
data = data.drop_duplicates()
```

```python
data.shape
```

```python
data['Gender'].value_counts()
```

```python
data.info()
```

```python
data[data['Age'].isnull()]
```

```python
data['Department'] = data['Department'].fillna('Not Available')
##data['Age'] = data['Age'].fillna(data['Age']).mean()
```

```python
data.isnull().sum()
```

```python
data
```

```python
## Join first and last name
data['Full Name']= data['First Name'] + data['Last Name']
```

```python
## Change Hire_Date to date time format
data['Hire_Date'] = pd.to_datetime(data['Hire_Date'])
```

```python
data
```

```python
data['year'] = data.Hire_Date.dt.year
data['current_year'] = datetime.now().year
```

```python
data
```

```python
employee_per_dept = data['Department'].value_counts()
plt.figure(figsize= (6,3))
plt.title('Department wise Employee')
plt.xlabel('Department')
plt.ylabel('Employee Count')
plt.xticks(rotation =0)
employee_per_dept.plot(kind = 'bar', color = 'skyblue')
plt.show()
```

```python
data.count()
```

```python
age_gender_data = data.groupby(['Age', 'Gender']).size().reset_index(name = 'count')
age_gender_data
```

```python
plt.figure(figsize= (10,6))
sns.lineplot(data = age_gender_data, x='Age', y='count', hue='Gender', marker= 'o')
plt.title('Employee Count By Age and Gender')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation =0)
plt.show()
```

```python
data.head()
```

```python
marital_age_gender_data = data.groupby(['Age', 'Gender', 'MaritalStatus']).size().reset_index(name = 'count')
marital_age_gender_data
```

```python
plt.figure(figsize= (6,3))
sns.lineplot(data = marital_age_gender_data, x='Age', y='count', hue='Gender', marker= 'o')
plt.title('Employee Count By Age, Gender and Marital Status')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation =0)
plt.show()
```

```python
marital_status_data = data.groupby(['Gender', 'MaritalStatus']).size().reset_index(name = 'count')
marital_status_data
```

```python
plt.figure(figsize= (6,3))
sns.lineplot(data = marital_status_data, x='Gender', y='count', hue='MaritalStatus', marker= 'o')
plt.title('Employee Count By Gender and Marital Status')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation =0)
plt.show()
```

```python
marital_data = data.groupby(['MaritalStatus']).size().reset_index(name = 'count')
marital_status = marital_data['MaritalStatus']
Count = marital_data['count']
```

```python
sns.set(style='whitegrid')
colors = sns.color_palette("colorblind")[0:len(marital_status)]
colors= ['darkblue', 'darkgreen', 'yellow']
plt.figure(figsize = (5,5))
plt.pie(Count, labels=marital_status, colors=colors, autopct='%1.1f%%', startangle= 140)
plt.axis('equal')
plt.xlabel('Strength of Marital Status', fontweight ='bold')
plt.show()
```
